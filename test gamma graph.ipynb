```python
from proximitygraphs import Gamma_Graph
from proximitygraphs import SetPoints
```

```python
vcolor = '#051650'
ecolor = '#0A2472'

uniform_sphere = SetPoints.uniform_sphere(n=10, dims = 2, seed=0)
uniform_sphere.draw(figsize=(6,6))
```

```python
H = Gamma_Graph(uniform_sphere, gamma0=0, gamma1=0)
H.draw2(title=True, details=True, figsize=(6,6), v_size=3, e_size=1, v_color=vcolor, e_color=ecolor, fontsize=10)
```

```python
H = Gamma_Graph(uniform_sphere, gamma0=1, gamma1=1)
H.draw2(title=True, details=True, figsize=(6,6), v_size=3, e_size=1, v_color=vcolor, e_color=ecolor, fontsize=10)
```

```python
H = Gamma_Graph(uniform_sphere, gamma0=-1, gamma1=-1)
H.draw2(title=True, details=True, figsize=(6,6), v_size=3, e_size=1, v_color=vcolor, e_color=ecolor, fontsize=10)
```

```python

gammas0 = [-1, -0.5, 0, 0.5, 1]
gammas1 = [-1, -0.5, 0, 0.5, 1]

for g0 in gammas0:
    for g1 in gammas1:
        if abs(g0) > abs(g1):
            continue
        H = Gamma_Graph(uniform_sphere, gamma0=g0, gamma1=g1)
        H.draw2(title=True, details=True, figsize=(6,6), v_size=3, e_size=1, v_color=vcolor, e_color=ecolor, fontsize=10)
```
