# Gamma Graph

Se muestran diferentes ejemplos de los gamma vecindarios para diferentes $n$ puntos, $\gamma_0$, $\gamma_1$.

```python
import proximitygraphs as pg
import time
```

```python
vcolor = '#051650'
ecolor = '#0A2472'

uniform_sphere = pg.SetPoints.uniform_sphere(n=512*24, dims = 2, seed=0)
uniform_sphere.draw(figsize=(6,6))
```

# Single Neighborhood, 512*24=12288 points

```python
start = time.time()
H = pg.Gamma_Graph(uniform_sphere, gamma0=0, gamma1=0, block_size=512)
H.draw2(title=True, details=True, figsize=(6,6), v_size=3, e_size=1, v_color=vcolor, e_color=ecolor, fontsize=10)
end = time.time()
print("Time taken: ", end - start)
```

```python
start = time.time()
H = pg.Gamma_Graph(uniform_sphere, gamma0=0, gamma1=0, block_size=16)
H.draw2(title=True, details=True, figsize=(6,6), v_size=3, e_size=1, v_color=vcolor, e_color=ecolor, fontsize=10)
end = time.time()
print("Time taken: ", end - start)
```
