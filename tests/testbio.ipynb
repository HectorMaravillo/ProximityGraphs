```python
import proximitygraphs as pg
import numpy as np
import matplotlib.pyplot as plt
```

# Physalum Polycephalum

```python
random_state = 0
points = pg.SetPoints.uniform_square(10, seed=random_state)
```

```python
points.draw(figsize=(8, 8), v_color='blue', v_size=20)
```

```python
ph = pg.PhysarumGraph(points, sources=[0], sinks=list(range(1, len(points.points))), steps=1000, dt=0.05, gamma=1.2, eps=1e-3, base_graph='delaunay')
ph.draw(
    figsize=(7, 7),
    e_size=[3],
    e_color="darkgreen",
    v_size=8
)
```

```python
import scipy.spatial.distance as dist

dist_matrix = dist.cdist(points.points, points.points)
print(dist_matrix)
```

```python
graph = ph.graph
```

```python
edge_list = graph.get_edgelist()
print(edge_list)
```
