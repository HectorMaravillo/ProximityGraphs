# `draw` method

Se verifica su funcionamiento.

```python
import proximitygraphs as pg
```

```python
uniform_sphere = pg.SetPoints.uniform_sphere(n=200, seed=0)
uniform_sphere.draw()
```

# Demo

```python
H = pg.Alpha_Shape(uniform_sphere, alpha=-5)
H.draw(details=True)
```

```python
H = pg.Alpha_Hull(uniform_sphere, alpha=-5, n_points_per_arc=50)
H.draw(details=True)
```

```python
H = pg.DelaunayG(uniform_sphere)
H.draw(details=True)
```

```python
H = pg.Beta_Skeleton(uniform_sphere, beta=2)
H.draw(details=True)
```

```python
H = pg.Convex_Hull(uniform_sphere)
H.draw(details=True)
```

```python
H = pg.MST(uniform_sphere)
H.draw(details=True)
```

```python
H = pg.Stepping_Stone(uniform_sphere, d=3, k=3)
H.draw(details=True)
```

```python
H = pg.NNG(uniform_sphere, k=2)
H.draw(details=True)
```

```python
H = pg.GG(uniform_sphere)
H.draw(details=True)
```

```python
H = pg.Elliptic_GabrielG(uniform_sphere)
H.draw(details=True)
```

```python
H.draw(
    v_kwargs={"marker": "x", "zorder": 3, "s": 20, "alpha": 0.6},
    e_kwargs={"linestyle": "--"},
    fig_kwargs={"facecolor": "none"},
    title_kwargs={"loc": "left", "fontsize": 16},
)
```
